```python
import os

for split in ["train", "valid", "test"]:
    split_path = os.path.join(DATASET_DRIVE, split)

    print(f"\n===== {split.upper()} =====")

    if not os.path.exists(split_path):
        print("❌ Missing folder")
        continue

    print(os.listdir(split_path))
```

```python
import os

IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
LABEL_EXTENSION = ".txt"

for split in ["train", "valid", "test"]:

    image_dir = os.path.join(DATASET_DRIVE, split, "images")
    label_dir = os.path.join(DATASET_DRIVE, split, "labels")

    images = [
        f for f in os.listdir(image_dir)
        if f.lower().endswith(IMAGE_EXTENSIONS)
    ]

    labels = [
        f for f in os.listdir(label_dir)
        if f.lower().endswith(LABEL_EXTENSION)
    ]

    print(f"\n{split.upper()}")
    print("Images :", len(images))
    print("Labels :", len(labels))
```

```python
import shutil
import os

LOCAL_PROJECT = "/content/MusaSense_YOLO11m"
LOCAL_DATASET = os.path.join(LOCAL_PROJECT, "dataset")

os.makedirs(LOCAL_PROJECT, exist_ok=True)

if os.path.exists(LOCAL_DATASET):
    shutil.rmtree(LOCAL_DATASET)

shutil.copytree(DATASET_DRIVE, LOCAL_DATASET)

print("Dataset copied successfully.")
print("Local dataset:", LOCAL_DATASET)
```

```python
import yaml
import os

with open(DATA_YAML_DRIVE, "r") as f:
    original_data = yaml.safe_load(f)

names = original_data["names"]

clean_data = {
    "path": LOCAL_DATASET,
    "train": "train/images",
    "val": "valid/images",
    "test": "test/images",
    "names": names
}

LOCAL_DATA_YAML = os.path.join(LOCAL_DATASET, "data.yaml")

with open(LOCAL_DATA_YAML, "w") as f:
    yaml.dump(clean_data, f, sort_keys=False)

print("Created:", LOCAL_DATA_YAML)

print("\nContents:")
with open(LOCAL_DATA_YAML, "r") as f:
    print(f.read())
```

```python
%%writefile /content/MusaSense_YOLO11m/train.py

from ultralytics import YOLO
import os


# ============================================================
# MusaSense - Pseudostem Weevil Detection
# YOLO11m Object Detection
# ============================================================

PROJECT_DIR = "/content/MusaSense_YOLO11m"

DATA_YAML = os.path.join(
    PROJECT_DIR,
    "dataset",
    "data.yaml"
)

RUNS_DIR = os.path.join(
    PROJECT_DIR,
    "runs"
)


# Load pretrained YOLO11m
model = YOLO("yolo11m.pt")


# Train
results = model.train(
    data=DATA_YAML,

    # Training
    epochs=100,
    imgsz=640,
    batch=-1,

    # GPU
    device=0,

    # Performance
    workers=2,
    amp=True,
    cache=False,

    # Optimization
    optimizer="auto",
    lr0=0.01,
    cos_lr=True,

    # Early stopping
    patience=40,

    # Reproducibility
    seed=42,
    deterministic=True,

    # Saving
    save=True,
    save_period=10,

    # Validation
    val=True,

    # Augmentation
    hsv_h=0.015,
    hsv_s=0.5,
    hsv_v=0.3,

    degrees=5.0,
    translate=0.1,
    scale=0.5,
    shear=2.0,
    perspective=0.0,

    fliplr=0.5,
    flipud=0.0,

    mosaic=1.0,
    mixup=0.05,

    close_mosaic=10,

    # Output
    project=RUNS_DIR,
    name="weevil_yolo11m_640",

    plots=True,
    verbose=True
)


print("\n========================================")
print("TRAINING COMPLETED")
print("========================================")
print("Best model:")
print(
    os.path.join(
        RUNS_DIR,
        "weevil_yolo11m_640",
        "weights",
        "best.pt"
    )
)
```
